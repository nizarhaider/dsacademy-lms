## Build the complete mental model

1. Fit imputation, scaling, and encoding inside each training fold to prevent leakage.
2. Cross-validation reports variation across folds, not one deceptively precise score.
3. Select hyperparameters using validation evidence and evaluate once on the untouched test set.
4. Persist preprocessing, model, schema, versions, parameters, metrics, and data identity together.

## Common mistakes
