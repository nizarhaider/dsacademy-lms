### Core mechanism

P(y=1|x)=1/(1+e^-(β₀+βᵀx)); a threshold turns probability into a class decision.

### Terms and symbols

- `probability`: A number from 0 to 1.
- `threshold`: The action boundary.
- `precision`: Correct positives among predicted positives.
- `recall`: Found positives among actual positives.


## Worked example: predict first

From TP=8, FP=2, FN=4, calculate precision=0.80, recall≈0.67, and F1≈0.73.

Before running the next cell:

1. Write the expected output.
2. Identify the input values.

````python
tp, fp, fn = 8, 2, 4
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * precision * recall / (precision + recall)
print(round(precision, 2), round(recall, 2), round(f1, 2))
````

## Build the complete mental model

1. Logistic regression converts a linear score into a probability with the sigmoid function.
2. Changing the threshold trades false positives against false negatives without retraining.
3. Accuracy can hide failure on a rare class; precision and recall expose different error costs.
4. Use MAE/RMSE/R² for regression and confusion-derived metrics for classification.

## Common mistakes

- Using accuracy for a rare but important class.
- Changing the threshold after viewing the final test result.
