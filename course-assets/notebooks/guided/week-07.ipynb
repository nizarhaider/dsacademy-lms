### Core mechanism

ŷ = β₀ + β₁x₁ + … + βₙxₙ; training chooses coefficients that minimize squared residuals.

### Terms and symbols

- `ŷ`: Predicted continuous value.
- `β₀`: Intercept.
- `βⱼ`: Coefficient for feature j.
- `xⱼ`: Feature value j.


## Worked example: predict first

For hours=3, β₀=40 and β₁=10, predict 70; with actual 76, residual=6 and squared error=36.

## Build the complete mental model

1. Ordinary least squares chooses coefficients that minimize the sum of squared residuals.
2. Closed-form solvers and gradient descent optimize the same objective through different procedures.
3. Check linearity, residual independence, roughly constant variance, and influential outliers.
4. Compare with a mean baseline; Ridge shrinks coefficients and Lasso can set some to zero.

## Common mistakes

- Calling RMSE an accuracy percentage even though it is an error measured in target units.
- Interpreting a coefficient causally or without considering feature scaling.
